```python
#create a list named products containing atleast 6 product names
products = ["Laptop", "Smartphone", "Headphones", "Smartwatch", "Tablet", "Camera"]

# create a tuple named sample_product that stores product name, price and category for one product 
sample_product = ("Laptop", 999.99, "Electronics")

```

```python
# print the 2nd and last product from the products list
print(products[1]) # 2nd 
print(products[-1]) # last 
```

```python
# append 2 new products into products list and print the updated list
products.append("Printer")
products.append("Monitor")
print(products)
```

```python
# convert sample_product into a list and change its price and then convert it back to tuple
sample_product_list = list(sample_product)
sample_product_list[1] = 899.99  # Change price
sample_product = tuple(sample_product_list)
```
