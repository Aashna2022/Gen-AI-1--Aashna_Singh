```python
price_dict = {'Laptop':999,
              'Smartphone':699,
              'Headphones':199,
              'Smartwatch':299,
              'Tablet':499,
              'Camera':799}

price_dict['Printer'] = 149
```

```python
print(price_dict)
```

```python
# update the laptop price to 899
price_dict['Laptop'] = 899
```

```python
# remove the product and also handle the case when its product name does not exist in the dictionary
if 'Printer' in price_dict:
    del price_dict['Printer']
else:
    print("Product not found in the dictionary.")
```

```python
print(price_dict)
```

```python
# print the avg price of all products 
total_price = 0
for product,price in price_dict.items():
    total_price += price
avg_price = total_price / len(price_dict)
print(f"The average price of all products is {avg_price}.")
```

```python
# print the max and min price 
max_price = max(price_dict.values())
min_price = min(price_dict.values())
```

```python
print(f'Min Price is -> {min_price}')
print(f'Max Price is -> {max_price}')
```
